## Card 5 — Put It All Together -- **PRE-BUILT — RUN THIS TO PLAY!**

This is Card 5's `while` loop from paper, now doing the real work: it keeps asking for
guesses until you either run out of wrong guesses or complete the word. It uses every
function you just built — `check_guess`, `get_display`, and your `EMOJI_STAGES`.

One extra trick: you can type a single letter, **or the whole word** for an instant
win. The `if len(guess) > 1:` line is what tells the two apart — it's just checking how
many characters you typed, the same `if` idea from Card 3.

Read through it before running — see if you can find where each earlier piece gets used.

```python
# 🔒 PRE-BUILT — run this cell to play!
guessed_letters = []
wrong_guesses = 0
word_display = get_display(secret_word, guessed_letters)

print("Guess a letter, or the whole word for an instant win!")

while wrong_guesses < MAX_WRONG_GUESSES and "_" in word_display:
    print(IMAGE_STAGES[wrong_guesses])
    print(word_display)

    guess = input("Guess a letter or the whole word: ").upper()

    if len(guess) > 1:
        # a whole-word guess
        if guess == secret_word:
            guessed_letters = list(secret_word)
            print("You guessed the whole word!")
        else:
            wrong_guesses = wrong_guesses + 1
            print("That's not the word!")
    else:
        guessed_letters.append(guess)
        if check_guess(guess, secret_word):
            print("Correct!")
        else:
            wrong_guesses = wrong_guesses + 1
            print("Wrong guess!")

    word_display = get_display(secret_word, guessed_letters)

if "_" not in word_display:
    print(WIN_MESSAGE.format(word=secret_word))
else:
    print(IMAGE_STAGES[-1])
    print(LOSE_MESSAGE.format(word=secret_word))
```
